```python

RED = "RED"
BLACK = "BLACK"


class Node:
    def __init__(self, data):
        self.data = data
        self.color = RED
        self.left = None
        self.right = None
        self.parent = None


class RedBlackTree:
    def __init__(self):
        self.NIL = Node(None)
        self.NIL.color = BLACK
        self.NIL.left = self.NIL
        self.NIL.right = self.NIL
        self.NIL.parent = self.NIL
        self.root = self.NIL

    # Left rotation
    def left_rotate(self, x):
        y = x.right
        x.right = y.left

        if y.left != self.NIL:
            y.left.parent = x

        y.parent = x.parent

        if x.parent == self.NIL:
            self.root = y
        elif x == x.parent.left:
            x.parent.left = y
        else:
            x.parent.right = y

        y.left = x
        x.parent = y

    # Right rotation
    def right_rotate(self, x):
        y = x.left
        x.left = y.right

        if y.right != self.NIL:
            y.right.parent = x

        y.parent = x.parent

        if x.parent == self.NIL:
            self.root = y
        elif x == x.parent.right:
            x.parent.right = y
        else:
            x.parent.left = y

        y.right = x
        x.parent = y

    # Insert a node
    def insert(self, data):
        z = Node(data)
        z.left = self.NIL
        z.right = self.NIL

        y = self.NIL
        x = self.root

        while x != self.NIL:
            y = x

            if data < x.data:
                x = x.left
            elif data > x.data:
                x = x.right
            else:
                print("Duplicate values are not allowed.")
                return False

        z.parent = y

        if y == self.NIL:
            self.root = z
        elif data < y.data:
            y.left = z
        else:
            y.right = z

        self.insert_fixup(z)
        return True

    # Balance after insertion
    def insert_fixup(self, z):
        while z.parent.color == RED:
            if z.parent == z.parent.parent.left:
                y = z.parent.parent.right

                if y.color == RED:
                    z.parent.color = BLACK
                    y.color = BLACK
                    z.parent.parent.color = RED
                    z = z.parent.parent
                else:
                    if z == z.parent.right:
                        z = z.parent
                        self.left_rotate(z)

                    z.parent.color = BLACK
                    z.parent.parent.color = RED
                    self.right_rotate(z.parent.parent)

            else:
                y = z.parent.parent.left

                if y.color == RED:
                    z.parent.color = BLACK
                    y.color = BLACK
                    z.parent.parent.color = RED
                    z = z.parent.parent
                else:
                    if z == z.parent.left:
                        z = z.parent
                        self.right_rotate(z)

                    z.parent.color = BLACK
                    z.parent.parent.color = RED
                    self.left_rotate(z.parent.parent)

        self.root.color = BLACK
        self.root.parent = self.NIL

    # Search for a value
    def search(self, data):
        x = self.root

        while x != self.NIL and x.data != data:
            if data < x.data:
                x = x.left
            else:
                x = x.right

        return x

    # Find minimum node
    def minimum(self, x):
        while x.left != self.NIL:
            x = x.left
        return x

    # Replace one subtree with another
    def transplant(self, u, v):
        if u.parent == self.NIL:
            self.root = v
        elif u == u.parent.left:
            u.parent.left = v
        else:
            u.parent.right = v

        v.parent = u.parent

    # Delete a node
    def delete(self, data):
        z = self.search(data)

        if z == self.NIL:
            print("Node not found.")
            return False

        y = z
        original_color = y.color

        if z.left == self.NIL:
            x = z.right
            self.transplant(z, z.right)

        elif z.right == self.NIL:
            x = z.left
            self.transplant(z, z.left)

        else:
            y = self.minimum(z.right)
            original_color = y.color
            x = y.right

            if y.parent == z:
                x.parent = y
            else:
                self.transplant(y, y.right)
                y.right = z.right
                y.right.parent = y

            self.transplant(z, y)
            y.left = z.left
            y.left.parent = y
            y.color = z.color

        if original_color == BLACK:
            self.delete_fixup(x)

        return True

    # Balance after deletion
    def delete_fixup(self, x):
        while x != self.root and x.color == BLACK:
            if x == x.parent.left:
                w = x.parent.right

                if w.color == RED:
                    w.color = BLACK
                    x.parent.color = RED
                    self.left_rotate(x.parent)
                    w = x.parent.right

                if (w.left.color == BLACK and
                        w.right.color == BLACK):
                    if w != self.NIL:
                        w.color = RED
                    x = x.parent
                else:
                    if w.right.color == BLACK:
                        w.left.color = BLACK
                        w.color = RED
                        self.right_rotate(w)
                        w = x.parent.right

                    w.color = x.parent.color
                    x.parent.color = BLACK
                    w.right.color = BLACK
                    self.left_rotate(x.parent)
                    x = self.root

            else:
                w = x.parent.left

                if w.color == RED:
                    w.color = BLACK
                    x.parent.color = RED
                    self.right_rotate(x.parent)
                    w = x.parent.left

                if (w.right.color == BLACK and
                        w.left.color == BLACK):
                    if w != self.NIL:
                        w.color = RED
                    x = x.parent
                else:
                    if w.left.color == BLACK:
                        w.right.color = BLACK
                        w.color = RED
                        self.left_rotate(w)
                        w = x.parent.left

                    w.color = x.parent.color
                    x.parent.color = BLACK
                    w.left.color = BLACK
                    self.right_rotate(x.parent)
                    x = self.root

        x.color = BLACK
        self.root.color = BLACK

    # Display tree structure
    def display(self, node=None, level=0, label="Root: "):
        if node is None:
            node = self.root

        if node == self.NIL:
            if level == 0:
                print("The tree is empty.")
            return

        print("    " * level + label +
              str(node.data) + " (" + node.color[0] + ")")

        self.display(node.left, level + 1, "L--- ")
        self.display(node.right, level + 1, "R--- ")


# Main menu
def main():
    tree = RedBlackTree()

    while True:
        print("\n===== RED-BLACK TREE MENU =====")
        print("1. Creation")
        print("2. Insertion")
        print("3. Deletion")
        print("4. Display")
        print("5. Exit")

        try:
            choice = int(input("Enter your choice: "))

            if choice == 1:
                tree = RedBlackTree()
                values = input(
                    "Enter initial elements separated by spaces: "
                ).strip()

                if values:
                    for value in values.split():
                        if tree.insert(int(value)):
                            print("Node inserted successfully.")

                print("Tree creation completed.")

            elif choice == 2:
                value = int(input("Enter the value to insert: "))

                if tree.insert(value):
                    print("Node inserted successfully.")

            elif choice == 3:
                value = int(input("Enter the value to delete: "))

                if tree.delete(value):
                    print("Node deleted successfully.")

            elif choice == 4:
                print("\nRed-Black Tree (R = Red, B = Black):")
                tree.display()

            elif choice == 5:
                print("Exiting the program.")
                break

            else:
                print("Invalid choice. Enter a number from 1 to 5.")

        except ValueError:
            print("Invalid input. Please enter integers.")


if __name__ == "__main__":
    main()

```
